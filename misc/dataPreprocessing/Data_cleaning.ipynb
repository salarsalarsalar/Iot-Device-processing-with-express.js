```python
# Data taken from "https://www.kaggle.com/datasets/fanbyprinciple/iot-device-identification"
import pandas as pd

# Load train and test CSVs (with headers)
df1 = pd.read_csv('iot_device_train.csv')
df2 = pd.read_csv('iot_device_test.csv')

# Merge datasets
df = pd.concat([df1, df2], ignore_index=True)

# Check shape and columns
print(df.shape)
print(df.columns.tolist())

# Check for null values
print(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
columns_to_keep = [
    'packet_size_avg', 'protocol', 'source_ip', 'destination_ip',
    'timestamp', 'packet_size_sum'
]

# Only keep these columns (if they exist in df)
filtered_df = df[[col for col in columns_to_keep if col in df.columns]]
```

```python
import pandas as pd
import numpy as np
import random

# Sample adjectives and new IoT-specific nouns
adjectives = ['smart', 'fast', 'tiny', 'silent', 'bright', 'sharp']
nouns = ['thermo', 'tracker', 'relay', 'gateway', 'cam', 'meter']

# Number of random names to generate
n = 10

# Function to create a random name
def generate_random_name():
    adj = random.choice(adjectives)
    noun = random.choice(nouns)
    number = random.randint(100, 999)
    return f"{adj}-{noun}-{number}"

# Create a DataFrame with random names
filtered_df['device_name'] = [generate_random_name() for _ in range(len(filtered_df))]
print(filtered_df)
```
